가상의 공유 자전거 운영사 RideOn은 지구별 하루 대여 건수를 미리 예측해 자전거 재배치 계획을 세우려고 합니다. 특히 비가 오는 날 예측 오차가 커지는지 확인하려고 합니다.

여러분은 운영 데이터 분석가로서 2025년 320일 동안 6개 지구의 일별 기록을 탐색하고, 날짜 순서를 지키는 분할로 수요 예측 회귀 모델을 만들고 평가해야 합니다.

- 분석 목표: 날짜와 지구별 대여 건수(`rides_count`) 예측

> 기온과 강우량은 운영 계획을 확정하기 직전에 받는 가상의 예보치입니다. `previous_week_same_day_rides`는 같은 지구의 7일 전 대여 건수로 예측 시점에 이미 알 수 있는 값입니다. 미래 날짜의 정보가 훈련에 섞이지 않도록 9번에서는 무작위 분할 대신 날짜 기준으로 나눕니다.
