## 문제 상황

가상의 스마트 빌딩 운영사 GridLab은 건물의 시간당 전력 사용량을 미리 예측해 에너지 운영 계획을 세우려고 합니다.

여러분은 에너지 데이터 분석가로서 건물 유형, 면적, 실내 인원, 실외 온도, 냉난방 운영 정보를 탐색하고 전력 사용량을 예측하는 회귀 모델을 만들고 평가해야 합니다.

- 분석 목표: 시간당 전력 사용량(`energy_kwh`, kWh) 예측
- 데이터: `building_energy.csv` (시간별 관측 1,650건, 14개 컬럼)
- 진행 순서: 데이터 탐색(1~5번) → 탐색용 전처리(6~8번) → 모델용 분할과 전처리(9번) → 모델 학습과 평가(10~14번)

> 한 행은 가상 건물의 1시간 관측입니다. `previous_hour_kwh`는 관측 직전 한 시간의 사용량으로 예측 시점에 알 수 있는 값입니다. 행들은 서로 독립적으로 만든 기록이라 무작위 분할을 씁니다. 실제 연속 시계열이라면 시간 순서를 지키는 검증이 필요합니다. 6~8번은 전체 데이터 탐색 연습이며, 9번은 원본 `energy_df`에서 다시 시작합니다.

### `building_energy.csv` 컬럼 설명

| 컬럼 | 타입 | 설명 |
|---|---|---|
| `reading_id` | 문자열 | 관측 고유 번호 (EN00001~EN01650). 식별용이라 모델 입력에는 쓰지 않습니다. |

| `occupied_people` | 정수 | 실내 인원 (명) |

| `insulation_grade` | 문자열 | 단열 등급: A(우수), B(보통), C(미흡) |
| `solar_generation_kwh` | 실수 | 태양광 발전량 (kWh). 야간에는 0 |
| `previous_hour_kwh` | 실수 | 직전 한 시간의 전력 사용량 (kWh). 예측 시점에 알 수 있는 값. 일부 결측 |
| `energy_kwh` | 실수 | 해당 시간 전력 사용량 (kWh). 예측 대상 |
