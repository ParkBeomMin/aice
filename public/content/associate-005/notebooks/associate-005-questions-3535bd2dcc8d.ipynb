- 분석 목표: 제안별 수락 여부(`responded`) 분류

- 진행 순서: 데이터 탐색(1~5번) → 탐색용 전처리(6~8번) → 모델용 분할과 전처리(9번) → 모델 학습과 평가(10~14번)

| `email_open_rate` | 실수 | 과거 30일 마케팅 이메일 열람률 (0~1). 이번 제안 이전 정보. 일부 결측 |

| `device_type` | 문자열 | 주 사용 기기: Desktop, Mobile, Tablet |
| `responded` | 정수 | 제안 수락 여부 (1: 수락, 0: 미수락). 예측 대상 |
