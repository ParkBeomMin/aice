## 문제 상황

가상의 제조 공장 MakeWorks는 생산 배치가 24시간 안에 품질 불량 판정을 받을지 미리 예측해 추가 검사를 배정하려고 합니다. 불량을 놓치는 일을 줄이는 것이 중요합니다.

여러분은 품질 데이터 분석가로서 생산 시작 전에 측정한 배치 정보를 탐색하고, 누설 없는 전처리로 불량 분류 모델을 만들고 교차 검증과 임계값 변경으로 평가해야 합니다.

- 분석 목표: 배치별 24시간 내 불량 여부(`defect_24h`) 분류
- 데이터: `production_batches.csv` (생산 배치 2,100건, 14개 컬럼)
- 진행 순서: 데이터 탐색(1~5번) → 탐색용 전처리(6~8번) → 모델용 분할과 전처리(9번) → 모델 학습과 평가(10~14번)

> 입력 컬럼은 모두 생산 시작 전에 알 수 있는 값입니다. `reject_count_prev_hour`는 직전 한 시간의 불량 건수, `precheck_score`는 생산 전 점검 점수입니다. 6~8번은 전체 데이터 탐색 연습이며, 9번과 13번은 원본에서 훈련 데이터(또는 각 폴드의 훈련 부분) 기준으로만 전처리합니다.
