## 문제 상황

가상의 로그인 보안 서비스 SecurePortal은 사용자 계정의 인증 시도 가운데 추가 확인이 필요한 이벤트를 미리 찾아내려고 합니다. 탐지된 이벤트에는 추가 인증을 요청할 계획입니다.

여러분은 보안 데이터 분석가로서 인증 이벤트를 탐색하고, 양성(`suspicious_login=1`)이 상대적으로 적은 분류 문제에서 재현율과 정밀도의 관계를 검토해야 합니다.

- 분석 목표: 인증 이벤트별 이상 로그인 여부(`suspicious_login`) 분류
- 데이터: `auth_events.csv` (인증 이벤트 1,800건, 14개 컬럼)
- 진행 순서: 데이터 탐색(1~5번) → 탐색용 전처리(6~8번) → 모델용 분할과 전처리(9번) → 모델 학습과 평가(10~14번)

> 6~8번은 전체 데이터로 하는 탐색 연습입니다. 9번은 반드시 원본 `auth_df`에서 다시 시작해 훈련 데이터 기준으로 전처리합니다. 판정 레이블은 합성 데이터의 가상 값이며 실제 계정 제한이나 보안 판단에 쓰지 않습니다.
